# FPsim Basic Sample

This notebook demonstrates the core functionality of FPsim:
1. Simple simulation with default parameters
2. Custom simulation with specific parameters
3. Multiple location examples
4. Basic plotting and analysis
5. Experiment framework usage

Run each cell to see FPsim in action!


## 1. Import Libraries


```python
import sciris as sc
import fpsim as fp
from fpsim import plotting as fp_plt
import matplotlib.pyplot as plt
import pandas as pd
```

## 2. Simplest Simulation


```python
# Run the simplest possible simulation
print("Running simplest simulation...")
sc.tic()
sim = fp.Sim()
sim.run()
elapsed = sc.toc()
if elapsed is not None:
    print(f"Completed in {elapsed:.2f} seconds")
else:
    print("Completed (timing not available)")
print(f"Final population: {sim.results['n_alive'][-1]:,.0f}")
```

## 3. Custom Simulation


```python
# Run a custom simulation with specific parameters
print("Running custom simulation (Kenya, 1000 agents)...")
pars = dict(
    location='kenya',
    n_agents=1000,
    start_year=2000,
    end_year=2020,
    exposure_factor=1.0
)

sc.tic()
sim_custom = fp.Sim(pars=pars)
sim_custom.run()
elapsed = sc.toc()
if elapsed is not None:
    print(f"Completed in {elapsed:.2f} seconds")
else:
    print("Completed (timing not available)")
print(f"Final population: {sim_custom.results['n_alive'][-1]:,.0f}")
```

## 4. Multiple Locations


```python
# Run simulations for different locations
locations = ['senegal', 'ethiopia']
results = {}

for location in locations:
    print(f"Running {location} simulation...")
    sc.tic()
    pars_loc = dict(
        location=location,
        n_agents=500,
        start_year=2000,
        end_year=2015
    )
    sim_loc = fp.Sim(pars=pars_loc)
    sim_loc.run()
    results[location] = sim_loc
    elapsed = sc.toc()
    if elapsed is not None:
        print(f"✓ {location.capitalize()} completed in {elapsed:.2f} seconds")
    else:
        print(f"✓ {location.capitalize()} completed (timing not available)")
    print(f"  Final population: {sim_loc.results['n_alive'][-1]:,.0f}")
```

## 5. Basic Plotting


```python
# Plot the main simulation results
fig = sim_custom.plot()
plt.show()
```

```python
# Plot calibration results
fig_calib = fp_plt.plot_calib(sim_custom)
plt.show()
```

## 6. Accessing Results


```python
# Access simulation results
print("Simulation Results:")
print(f"Final population size: {sim_custom.results['n_alive'][-1]:,.0f}")
print(f"Total deaths: {sim_custom.results['cum_deaths'][-1]:,.0f}")
print(f"Urban population: {sim_custom.results['n_urban'][-1]:,.0f}")
print(f"\nAvailable result keys: {list(sim_custom.results.keys())}")
```

```python
# Create a simple comparison plot
fig, ax = plt.subplots(figsize=(10, 6))

for location, sim in results.items():
    ax.plot(sim.results['timevec'], sim.results['n_alive'], label=location.capitalize(), linewidth=2)

ax.set_xlabel('Time (months)')
ax.set_ylabel('Population Size')
ax.set_title('Population Growth Comparison')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 7. Experiment Framework


```python
# Run experiment framework (if available)
try:
    print("Running experiment framework...")
    sc.tic()
    exp = fp.Experiment()
    exp.run()
    elapsed = sc.toc()
    if elapsed is not None:
        print(f"✓ Experiment completed in {elapsed:.2f} seconds")
    else:
        print("✓ Experiment completed (timing not available)")
    
    # Show experiment summary
    df = exp.summarize()
    print("\nExperiment Summary:")
    if hasattr(df, 'to_string'):
        print(df.to_string())
    else:
        print(df)
    
except Exception as e:
    print(f"⚠ Experiment failed: {e}")
    print("   (This may require additional setup)")
```
